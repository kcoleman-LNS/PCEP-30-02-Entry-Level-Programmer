```python
#Keywords break and continue
#Break: exit loop and next instruction outside loop (commonly used with if statement)
#Continue: escape the CURRENT interation, often used with if statements
```

```python
while True: #Infinite loop, so we use the break
    name = input('Enter your name or EXIT to close the program: ')
    if (name == 'EXIT'):
        break
    print('Hello', name)
```

```python
for i in range(1,21):
    if i% 5 == 0: #Check for numbers divisible by 5 and exclude
        continue
    print(i)
```
