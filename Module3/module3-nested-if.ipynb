```python
user_age = int(input('Enter your age: '))
user_country = input('Enter your country: ')

if not user_country == 'China' and user_age > 18 or \
        user_country == 'Taiwan':
    print('Sorry, you do not qualify.')
else:
    print('You do qualify!')
```

```python
answer_a = input('Do you like travelling? y/n: ')
if answer_a == 'y':
    answer_b = input('Do you like Asia? y/n: ')
    if answer_b == 'y':
        print('Excellent, you can win a ticket to Japan!')
    else:
        print('That is unfortunate! You are missing out! ')
else:
    print('You do not like travelling! Boo!')
```

```python
## Nested Exercise 5
purchase = int(input('How many days ago have you purchased the item? '))

if purchase < 4:
    item_use = input('Have you used the item at all [y/n]? ')
    if item_use == 'n':
        item_broken = input('Has the item broken down on it\'s own [y/n]? ')
        if item_broken == 'y':
                print('You can get a refund.')
else:
    print('You cannot get a refund.')
```

```python
## Using Boolean operators
purchase_days_ago = int(input('How many days ago have you purchased the item? '))
is_used = input('Have you used the item at all [y/n]? ')
is_broken = input('Has the item broken down on its own [y/n]? ')

if(is_broken == 'y' or (purchase_days_ago <= 10 and is_used == 'n')):
  print('You can get a refund.')
else:
  print('You cannot get a refund.')
```
