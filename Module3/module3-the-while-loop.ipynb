```python
secret_number = 42
print('''
SECRET NUMBER GUESSING GAME!
''')
user_input = int(input('Guess the secret number between 1-100: '))
while user_input != secret_number:
    print('Incorrect, try again!')
    user_input = int(input('Guess the secret number between 1-100: '))
print('You have found the secret number!')
```
