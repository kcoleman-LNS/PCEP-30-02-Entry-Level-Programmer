```python
# Boolean operators; and, or, not
```

```python
user_age = int(input('Enter your age: '))
user_country = input('Enter your country: ')
if user_age >= 18 and user_country == 'China':
    print('You are eligible for Chinese/USA work visa.')
else:
    print('Sorry, you do not qualify.')
```

```python
user_age = int(input('Enter your age: '))
user_country = input('Enter your country: ')

if user_age >= 18 and user_country == 'China' or 'Taiwan' or 'Japan':
    print('You are eligible for Eastern Country/USA work visa.')
else:
    print('Sorry, you do not qualify.')
```

```python
# Not negates the boolean operator, or turns it around
# The order of operations for boolean; 1. not, 2. and, 3. or
```

```python
user_age = int(input('Enter your age: '))
user_country = input('Enter your country: ')

if not user_country == 'China' and user_age > 18 or \
        user_country == 'Taiwan':
    print('Sorry, you do not qualify.')
else:
    print('You do qualify!')
```
