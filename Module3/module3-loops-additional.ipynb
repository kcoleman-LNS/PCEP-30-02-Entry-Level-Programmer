```python
#Nested loops: times table example
for a in range(1,6):
    for b in range(1,6):
        print(a, 'x', b, '=', a*b)
```

```python
# Loops with else statements: rarely used
# The else branch of a loop is ALWAYS executed exactly once (exception: break statement)
i = 2
while i < 5:
    print(i)
    i += 1
else:
    print('else:', i)
```
