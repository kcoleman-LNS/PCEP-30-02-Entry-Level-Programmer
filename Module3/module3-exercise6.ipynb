```python
py_date = 1994
while True:
    release_date = int(input('When was Python 1.0 released? '))
    if (release_date == py_date):
        print('Correct!')
        break
    if (release_date < py_date):
        print('It was later than that! '
                'When was Python 1.0 released? ')
    elif (release_date > py_date):
        print('It was earlier than that! '
                'When was Python 1.0 released? ')

```

```python
# More simple example shown
while True:
    answer = int(input('When was Python 1.0 released? '))
    if answer > 1994:
        print('It was earlier than that!')
    elif answer < 1994:
        print('It was later than that!')
    else:
        print('Correct!')
        break
```
