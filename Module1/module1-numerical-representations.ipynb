```python
#Underscores are available, they don't change the number but they are good for more readability because you can't use commas, periods, or spaces.
```

```python
# scientific notation
# 3e4 = 3E4 = 3 * 10000
print(0.0000000000000000000000000000003)
```

```python
# octal numbers (rarely used)
# start with 0O 0o (zero, oh, or zero, small oh)
print(0o123)
```

```python
#hexadecimal numbers
# start with 0X or 0x
print(0x123)
```
