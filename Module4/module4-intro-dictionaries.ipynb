```python
# Used to store Key Value Pairs
# Very Fast

emails = {
    'Anne Stahl': 'astalh@gmail.com',
    'Peter Small': 'peters@yandex.com',
    'Mark Steel': 'mark@steel.com'
}

print(emails['Anne Stahl'])
```

```python
print(emails['Peter Small']) # Show me the value for the key 'Peter Small'
```
