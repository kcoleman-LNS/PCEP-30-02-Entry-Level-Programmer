```python
# Tuple elements in Lists
# Lists elements in tuples
# NOTE about [] and () when it comes to indexing. Python uses square brackets [] for indexing across all sequence types—including tuples—because parentheses () are reserved for calling functions, and separating these two operations prevents syntax ambiguity.
```

```python
city_1 = ('London', 'UK', 8.98)
city_2 = ('Canberra', 'Australia', 0.4)
city_3 = ('Algiers', 'Algeria', 3.9)
# capitals = [('London', 'UK', 8.98), ('Canberra', 'Australia', 0.4), ('Algiers', 'Algeria', 3.9)]
capitals = [city_1, city_2, city_3]
print(capitals)
```

```python
for capital in capitals:
    print('Name:', capital[0], ', Country:', capital[1], ', Population:', capital[2])
    # Prints each element in the list, printing by selecting their index within the tuples
```

```python
# List inside tuple
user_data = ('John', 'American', 1964, [77.0, 78.2, 77.5]) # (immutable tuple data [mutable list data])
```

```python
user_data[3].append(99.0)
```
