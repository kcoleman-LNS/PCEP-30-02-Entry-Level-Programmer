```python
grades = {}
```

```python
# Add new elements
grades['John'] = 'A-'
```

```python
grades['Anne'] = 'B-'
```

```python
print(grades)
```

```python
grades.update({'John': 'F'})
```

```python
print(grades)
```

```python
len(grades)
```

```python
# Key Check
if 'John' in grades:
    print('John got:', grades['John'])
```

```python
del grades['John']
print(grades)
```

```python
# Iterate a dictionary; prior to 3.6 dictionaries were not ordered, but they are now by default

# Add new elements
grades['John'] = 'A-'
grades['Anne'] = 'B-'
print(grades)
```

```python
for el in grades:
    print(el)
```

```python
for el in grades.keys():
    print(el)
```

```python
for el in grades.values():
    print(el)
```

```python
for person, grade in grades.items():
    print(person, 'got', grade)
```
