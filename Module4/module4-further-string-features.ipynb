```python
# Strings are sequences
fav_band = 'Sleep Token'
print(fav_band[:])
```

```python
#  You can't use indexing to change a single character in a string
fav_band[6] = 'M'
```

```python
# No append, insert, sort
# But you can do a lot with them; upper and lower for instance
print(fav_band.upper())
```

```python
# isnumeric can be used to check if user input is number
user_number = input('Enter a number: ')
if user_number.isnumeric():
    print('Thank you for providing a number. That\'s a great number')
else:
    print('That\'s not a number')
```
