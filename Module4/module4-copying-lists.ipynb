```python
# These lists are modified at the same time, so we must use slicing to create a real new list
list_original = [1, 2, 3, 4, 5]
list_new = list_original
list_original[0] = -5
print('Original:', list_original, '\nNew:', list_new)
```

```python
list_original = [1, 2, 3, 4, 5]
list_new = list_original[:] #Using slicing to create a new list
list_original[0] = -5
print('Original:', list_original, '\nNew:', list_new)
```

```python
list_original = [1, 2, 3, 4, 5]
list_new = list_original[:2] #Using slicing to create a new list
list_original[0] = -5
print('Original:', list_original, '\nNew:', list_new)
```
