```python
invited_guests = ['Kate', 'Adrian', 'Adam', 'Kerry', 'Joe', 'Anne']
name = input("What is your name? ")
if name in invited_guests:
    print('Welcome!')
else:
    print('Sorry, you were not invited.')
```

```python
invited_guests = ['Kate', 'Adrian', 'Adam', 'Kerry', 'Joe', 'Anne']
name = input("What is your name? ")
if name not in invited_guests:
    print('Welcome!')
else:
    print('Sorry, you were not invited.')
```
