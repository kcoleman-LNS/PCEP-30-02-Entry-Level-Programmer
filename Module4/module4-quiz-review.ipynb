```python
list_a = [1, 2, 3]
list_b = list_a[-2:-1] #starts at -2 from rear(inclusive) and ends at
list_c = list_a[-1:-2]

# list b is 3, 2
# list c is 2, 1
```

```python
print(list_a)
print(list_b)
print(list_c)
```

```python
list_a = [1, 2, 3]
list_b = list_a[-2:] #starts at -2 from rear(inclusive) and ends at
list_c = list_a[-1:]
```

```python
print(list_a)
print(list_b)
print(list_c)
```
