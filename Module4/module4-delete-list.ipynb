```python
city_1 = 'San Andreas'
city_2 = 'Madrid'
city_3 = 'Kyoto'
city_4 = 'Tokyo'

top_cities = [city_1, city_2, city_3, city_4]
print(top_cities)
```

```python
del top_cities[2]
print(top_cities)
```

```python
# The elements shift, the index doesn't hold an empty value
```

```python
top_cities = [city_1, city_2, city_3, city_4]
print(top_cities)
```

```python
del top_cities[:]
print(top_cities)
```
