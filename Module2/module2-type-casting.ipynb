```python
temp_f = input('Enter the temperature today in Fahrenheit: ')
temp_c = float(temp_f) / 1.8 - 32
temp_statement = str(temp_c) + ' degrees Fahrenheit equals ' + str(temp_f) + ' degrees Celsius'
print(temp_statement)
```
