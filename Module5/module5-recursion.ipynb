```python
# No recursion, iterative
def get_factorial(number):
    factorial = 1
    for x in range(1, number +1):
        factorial *= x
    return factorial

get_factorial(12)
```

```python
# With recursion
def get_factorial_rec(number):
    if number == 1:
        return 1
    return number * get_factorial_rec(number - 1) #But we need a stop condition after this or it will not end!

```

```python
get_factorial_rec(6)
# 6 * get_factorial(5) -> it continues down until 1
```
