```python
try:
    value = int(input('Enter an integer: '))
    print('The inverse of', value, 'is', 1/value)
except:
    print('You did not provide a number, so I will not calculate the inverse')
```

```python
# But what about zero? A special corner case
try:
    value = int(input('Enter an integer: '))
    print('The inverse of', value, 'is', 1/value)
except:
    print('You did not provide a number, so I will not calculate the inverse')
```

```python
# Sometimes we need to KNOW what the exception is
value = int(input('Enter an integer: '))
print('The inverse of', value, 'is', 1/value)
```

```python
# Of except branches is not limited, but they can only be deployed once before the program is terminated
try:
    value = int(input('Enter an integer: '))
    print('The inverse of', value, 'is', 1/value)
except ValueError:
    print('You did not provide a number, so I will not calculate the inverse')
except ZeroDivisionError:
    print('You provided 0 and division is by 0 is not possible, try again')
```
