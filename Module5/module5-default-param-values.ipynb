```python
def print_letter_count(text='superman is dead', letter='e'): # two parameters separated by commans, if you invoke, you need to provide both provided.
    counter = 0
    for char in text:
        if char == letter:
            counter += 1
    print('Number of', letter, 'is', counter)
```

```python
print_letter_count('The stores are closing because of the economy')
```

```python
print_letter_count()
```

```python
print_letter_count(text='superman is dead', letter)
```

```python
print_letter_count(letter, text='superman is dead')
```
