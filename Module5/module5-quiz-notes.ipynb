```python
# Because the global keyword is inside the function, it modifies the global b variable, even outside the function
def func(a):
    global b
    b = a + a
    return b

b = 10
print(func(13))
print(b)
```

```python

def func(x):
  if x == 0:
    return 0
  return x + func(x - 1) # Recursive function

print(func(3))
```

```python
# Shadowing
def func(x):
  x += 1 # interation
  return x

x = 5 # this gets iterated after because it's called during y, so  it gets it's +1 before the +2
y = func(x + 2)
print(x, y)
```

```python
# Generators! because it uses yield! Range, remainder operator, keyword end
def numerical():
  for i in range(10):
    yield i % 2

for x in numerical():
  print(x, end='-')
```
