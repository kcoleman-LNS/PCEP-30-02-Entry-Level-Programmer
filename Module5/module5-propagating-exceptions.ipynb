```python
def get_day(user_info):
    day = int(input('What is your birth day? '))
    user_info.append(day)

def get_month(user_info):
    month = int(input('What is your birth month (1-12)? '))
    user_info.append(month)

def get_year(user_info):
    year = int(input('What is your birth year? '))
    user_info.append(year)

#we can use a single try...except block within the function that houses the other functions
def get_user_bday(user_info):
    try:
        get_day(user_info)
        get_month(user_info)
        get_year(user_info)
    except ValueError:
        print('So your birth day is', user_info)

print('Hi, I will collect some info about your birth day!')
user_info = []
get_user_bday(user_info)
```
