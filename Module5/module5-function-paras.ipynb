```python


def get_average(input_numbers): # Parameters are special variables, they only can be found inside this function
    sum = 0.0
    for number in input_numbers:
        sum += number
    average = sum / len(input_numbers)
    print(average)

get_average([5.0, 3.5, 7.8, 9.9, 10.0]) # This list is an argument being passed
```

```python
# Requires the specific data type that is defined already
def concatenate(sentence):
    line = ''
    for character in sentence:
        line += character
    print(line)

concatenate(['Stars ', 'Align ', 'When ', 'You ', 'Observe ', 'Them!'])
```

```python
def print_letter_count(text, letter): # two parameters separated by commans, if you invoke, you need to provide both provided.
    counter = 0
    for char in text:
        if char == letter:
            counter += 1
    print('Number of', letter, 'is', counter)
```

```python
print_letter_count('Stars Align When You Observe Them!', 'e')
```

```python
print_letter_count('Stars Align When You Observe Them!', 'o')
```

```python
print_letter_count('e', 'Stars Align When You Observe Them!')
```

```python
# This are positional arguments, it matters where you put them, and this is in most languages
```

```python
# Keyword arguments 'named arguments', this can be used in lieu of positions
print_letter_count(text='Stars Align When You Observe Them!', letter='e')
```
