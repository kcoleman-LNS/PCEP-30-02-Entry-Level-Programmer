```python
# Shadowing, calling a function with a variable defined, but then you can create a second, temporary variable before it switches back
def show_truth():
    mysterious_var = 'New Surprise!' # SHADOW variable
    print(mysterious_var)

mysterious_var = 'Surprise!' # GLOBAL variable
print(mysterious_var)
show_truth()
print(mysterious_var)
```

```python
def show_truth():
    global mysterious_var # Don't use shadowing (don't create a second temporary variable); but
    print(mysterious_var)

mysterious_var = 'Surprise!' # GLOBAL variable
print(mysterious_var)
show_truth()
print(mysterious_var)
```

```python
def show_truth():
    mysterious_var.append('New Surprise!') # Changes the global element
    print(mysterious_var)

mysterious_var = ['Surprise!']
print(mysterious_var)
show_truth()
print(mysterious_var)
```
