```python
def get_number():
    for i in range(1, 4):
        yield i
```

```python
print(get_number())
```

```python
# Store generator in a variable, then ask for next value with next function
generator = get_number()
print(next(generator))
print(next(generator))
print(next(generator))
```

```python
def count_up_to(max_num):
    count = 1
    while count <= max_num:
        yield count  # Pauses here and returns 'count'
        count += 1

counter = count_up_to(5)
```

```python
def count_up_to(max_num):
    count = 1
    while count <= max_num:
        yield count  # Pauses here and returns 'count'
        count += 1
    if count > max_num:
        print('End of count.')
        return
counter = count_up_to(3)
```

```python
print(next(counter))
print(next(counter))
print(next(counter))
print(next(counter))
```
